Precision@K measures how many of the top K content items identified by the model actually match the defined review-priority proxy. A higher Precision@K means the ranked list is more useful because more of the items sent for limited human review are relevant to the review goal.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could flag content using one threshold, such as a large drop in clicks or impressions. However, search performance depends on multiple signals at the same time, including recent and previous performance, impressions, clicks, content characteristics, and search context.

An ML scoring approach can combine these signals and rank content items by their overall pattern instead of relying on one arbitrary threshold. This makes the output more useful for prioritizing limited human review.

The score is decision-support, not an automatic decision to refresh or change a page.
